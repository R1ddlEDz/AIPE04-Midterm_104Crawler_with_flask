```python
import requests
from bs4 import BeautifulSoup as bs
import re
from pprint import pprint
custom_headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/150.0.0.0 Safari/537.36",
    "Referer" : "https://104.com.tw"
}

query = ''

params = {
    "ro" : "1", #0全部職缺 1正2兼
    "keyword" : "python", #關鍵字
    "area" : "6001001000", #地區(台北市)
    "jobexp" : "1" #工作經驗 1年以下
}

#url = f'https://www.104.com.tw/jobs/search/api/jobs?area={area}&jobsource=index_s&keyword={key_word}&mode=s&order=15&page=1&pagesize=20{job_exp}'
url = "https://www.104.com.tw/jobs/search/api/jobs?"
res = requests.get(url,headers=custom_headers,params=params)



print(res.status_code)

data = res.json()
pprint(data, sort_dicts=False)
```

```python
keep_keys = [
    'appearDate', 'applyCnt', 'custName', 'jobName', 'description', 
    'jobAddrNoDesc','jobAddress', 'link', 'salaryLow'
]
target_dict = data['data'][0]
target_list = data['data']

filtered_list = [
    {
        k: (
        item[k].get('job') if k == 'link' and isinstance(item[k], dict) else
        (item[k].replace('\n', '').replace('\r','') if k == 'description' and isinstance(item[k], str) else item[k]))
        for k in keep_keys if k in item
    }
    for item in target_list
] #清洗資料 只保留keep_keys要求的鍵值，再加上三元運算，如果key=description則觸發replace"\n" 剩下的則原封不動放入filtered_list


pprint(filtered_list, sort_dicts=False)
#pprint(data['data'][0]['applyCnt'], sort_dicts=False)
```
